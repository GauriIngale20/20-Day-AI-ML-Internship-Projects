## 5. Data Cleaning

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing numerical values with median
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
df['experience'] = pd.to_numeric(df['experience'], errors='coerce')

df['age'] = df['age'].fillna(df['age'].median())
df['salary'] = df['salary'].fillna(df['salary'].median())
df['experience'] = df['experience'].fillna(df['experience'].median())

# Fill missing categorical values with mode
for col in ['department', 'city']:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# Remove duplicate records

df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Updated dataset shape:", df.shape)
```

## 6. Data Type Correction

```python
# Convert joining_date into datetime format

df['joining_date'] = pd.to_datetime(df['joining_date'], errors='coerce')

# Display updated data types
print(df.dtypes)
```

## 7. Data Filtering

```python
# Display employees with salary greater than 50000

high_salary = df[df['salary'] > 50000]

print("Employees with salary greater than 50000:")
display(high_salary.head(10))
```

## 8. Grouping Data

```python
# Group employees by department

department_summary = df.groupby('department').agg(
    Total_Employees=('employee_id', 'count'),
    Average_Salary=('salary', 'mean'),
    Average_Age=('age', 'mean')
)

display(department_summary)
```

## 9. Joining DataFrames

```python
# Create a department information DataFrame

department_info = pd.DataFrame({
    'department': ['IT', 'HR', 'Finance', 'Sales'],
    'department_head': ['Rahul', 'Priya', 'Amit', 'Sneha']
})

# Merge employee data with department information

merged_df = pd.merge(df, department_info, on='department', how='left')

display(merged_df.head(10))
```

## 10. Summary Statistics

```python
# Generate summary statistics

print("Total Employees:", len(df))
print("Average Salary:", df['salary'].mean())
print("Maximum Salary:", df['salary'].max())
print("Minimum Salary:", df['salary'].min())

display(df.describe())
```

## 11. Exporting Cleaned Dataset

```python
# Export cleaned dataset

from pathlib import Path

output_path = Path("../reports/cleaned_employees.csv")
df.to_csv(output_path, index=False)

print("Cleaned dataset exported successfully!")
print("File location:", output_path)
```

## 12. Final Data Validation and Cleaning

```python
# Standardize department and city names

df['department'] = df['department'].str.strip().str.upper()
df['city'] = df['city'].str.strip().str.title()

# Handle invalid age values
df.loc[(df['age'] < 18) | (df['age'] > 60), 'age'] = np.nan
df['age'] = df['age'].fillna(df['age'].median())

# Handle invalid salary values
df.loc[df['salary'] <= 0, 'salary'] = np.nan
df['salary'] = df['salary'].fillna(df['salary'].median())

# Display final validation results
print("Final Missing Values:")
print(df.isnull().sum())

print("\nFinal Department Categories:")
print(df['department'].unique())
```

## 13. Data Quality Summary

```python
# Load original dataset for comparison

original_df = pd.read_csv("../data/employees.csv")

# Generate data quality summary

summary = f"""
# Data Quality Summary Report

## Dataset Information
- Original Rows: {len(original_df)}
- Original Columns: {len(original_df.columns)}
- Cleaned Rows: {len(df)}
- Cleaned Columns: {len(df.columns)}

## Data Quality Checks
- Missing Values in Original Dataset: {original_df.isnull().sum().sum()}
- Duplicate Rows in Original Dataset: {original_df.duplicated().sum()}
- Duplicate Rows After Cleaning: {df.duplicated().sum()}
- Missing Values After Cleaning: {df.isnull().sum().sum()}

## Cleaning Operations
1. Handled missing numerical values.
2. Handled missing categorical values.
3. Removed duplicate records.
4. Corrected numerical data types.
5. Converted joining_date into datetime format.
6. Standardized department and city names.
7. Handled invalid age and salary values.

## Conclusion
The dataset was inspected, cleaned, and analyzed using Pandas and NumPy.
The cleaned dataset is ready for further analysis.
"""

print(summary)
```

```python
# Save data quality summary as Markdown file

report_path = Path("../reports/data_quality_summary.md")

report_path.write_text(summary, encoding="utf-8")

# Export the final cleaned dataset again
df.to_csv("../reports/cleaned_employees.csv", index=False)

print("Data Quality Summary saved successfully!")
print("Final cleaned dataset saved successfully!")
```
